 A. Список квадратов Это первая задача в блоке, и она поможет освоиться с базовой конструкцией списочных выражений в Python. Такие выражения позволяют лаконично и эффективно создавать списки по заданному правилу.

Вашему решению будут даны две переменные `a` и `b`.  
Напишите одно списочное выражение, которое формирует список квадратов всех целых чисел от [a,b][a,b] включительно.

## Примечание

В решении не должно быть ничего, кроме списочного выражения.

```python
a = -8
b = 11
squares = [x ** 2 for x in range(a, b + 1)]
print(squares)
```

Вашему решению даны переменные `a` и `b`. Сформируйте список квадратов всех целых чисел от `a` до `b` включительно — в прямом или обратном порядке, в зависимости от того, какое значение больше.
